```python
import tensorflow as tf
import numpy as np
print('TensorFlow version:', tf.__version__)
```

```python
# 0D tensor (scalar)
scalar = tf.constant(7)
print('0D Tensor (scalar):', scalar.numpy(), 'Shape:', scalar.shape)
```

```python
# 1D tensor (vector)
vector = tf.constant([1, 2, 3, 4])
print('1D Tensor (vector):', vector.numpy(), 'Shape:', vector.shape)
```

```python
# 2D tensor (matrix)
matrix = tf.constant([[1,2,3],[4,5,6]])
print('2D Tensor (matrix):\\n', matrix.numpy(), '\\nShape:', matrix.shape)
```

```python
# 3D tensor (cube / single image example)
image = tf.constant([[[255,0,0],[0,255,0]],
                     [[0,0,255],[255,255,0]]])
print('3D Tensor (image):\\n', image.numpy(), '\\nShape:', image.shape)
```

```python
# 4D tensor (batch of images)
batch = tf.constant([
  [[[255,0,0],[0,255,0]], [[0,0,255],[255,255,0]]],
  [[[0,255,255],[255,0,255]], [[128,128,128],[0,0,0]]]
])
print('4D Tensor (batch of images):\\n', batch.numpy(), '\\nShape:', batch.shape)
```
