```python
import tensorflow as tf
import numpy as np
print('TensorFlow version:', tf.__version__)
```

```python
# 0D tensor (scalar)
scalar = tf.constant(7)
print('0D Tensor (scalar):', scalar.numpy(), 'Shape:', scalar.shape)
```

```python
# 1D tensor (vector)
vector = tf.constant([1, 2, 3, 4])
print('1D Tensor (vector):', vector.numpy(), 'Shape:', vector.shape)
```

```python
# 2D tensor (matrix)
matrix = tf.constant([[1,2,3],[4,5,6]])
print('2D Tensor (matrix):\\n', matrix.numpy(), '\\nShape:', matrix.shape)
```

```python
# 3D tensor (cube / single image example)
image = tf.constant([[[255,0,0],[0,255,0]],
                     [[0,0,255],[255,255,0]]])
print('3D Tensor (image):\\n', image.numpy(), '\\nShape:', image.shape)
```

```python
# 4D tensor (batch of images)
batch = tf.constant([
  [[[255,0,0],[0,255,0]], [[0,0,255],[255,255,0]]],
  [[[0,255,255],[255,0,255]], [[128,128,128],[0,0,0]]]
])
print('4D Tensor (batch of images):\\n', batch.numpy(), '\\nShape:', batch.shape)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def show_tensor_values(tensor, name):
    shape = tensor.shape
    print(f"{name} shape: {shape}")
    
    # Convert to numpy
    t = tensor.numpy()
    
    if len(shape) == 0:
        plt.text(0.5, 0.5, str(t), fontsize=18, ha='center')
        plt.title(f"{name} (0D Scalar)")
        plt.axis('off')
        plt.show()
    elif len(shape) == 1:
        plt.bar(range(shape[0]), t)
        plt.title(f"{name} (1D Vector)")
        plt.show()
    elif len(shape) == 2:
        plt.imshow(t, cmap='viridis')
        plt.title(f"{name} (2D Matrix)")
        plt.colorbar()
        plt.show()
    elif len(shape) == 3:
        fig, axes = plt.subplots(1, shape[0])
        for i in range(shape[0]):
            axes[i].imshow(t[i], cmap='viridis')
            axes[i].set_title(f"Layer {i}")
        fig.suptitle(f"{name} (3D Tensor)")
        plt.show()
    elif len(shape) == 4:
        batch_size = shape[0]
        fig, axes = plt.subplots(1, batch_size)
        for i in range(batch_size):
            axes[i].imshow(t[i,:,:,0], cmap='viridis')  # show first channel for simplicity
            axes[i].set_title(f"Sample {i}")
        fig.suptitle(f"{name} (4D Batch)")
        plt.show()

# Run it
show_tensor_values(scalar, "Scalar")
show_tensor_values(vector, "Vector")
show_tensor_values(matrix, "Matrix")
show_tensor_values(image, "3D Image")
show_tensor_values(batch, "4D Batch")
```
