```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
print('TensorFlow version:', tf.__version__)
```

```python
# 0D Tensor (scalar)
scalar = tf.constant(7)
print('0D Tensor (scalar):', scalar.numpy(), 'Shape:', scalar.shape)
plt.text(0.5, 0.5, str(scalar.numpy()), fontsize=30, ha='center')
plt.title('0D Tensor (Scalar)')
plt.axis('off')
plt.show()
```

```python
# 1D Tensor (vector)
vector = tf.constant([1,2,3,4])
print('1D Tensor (vector):', vector.numpy(), 'Shape:', vector.shape)
plt.bar(range(vector.shape[0]), vector.numpy())
plt.title('1D Tensor (Vector)')
plt.show()
```

```python
# 2D Tensor (matrix)
matrix = tf.constant([[1,2,3],[4,5,6]])
print('2D Tensor (matrix):\n', matrix.numpy(), '\nShape:', matrix.shape)
plt.imshow(matrix.numpy(), cmap='viridis')
for i in range(matrix.shape[0]):
    for j in range(matrix.shape[1]):
        plt.text(j, i, str(matrix.numpy()[i,j]), ha='center', va='center', color='white')
plt.title('2D Tensor (Matrix)')
plt.colorbar()
plt.show()
```

```python
# 3D Tensor (Image-like)
image = tf.constant([[[255,0,0],[0,255,0]],
                     [[0,0,255],[255,255,0]]], dtype=tf.float32)
print('3D Tensor (image):\n', image.numpy(), '\nShape:', image.shape)
fig, axes = plt.subplots(1, image.shape[0])
for i in range(image.shape[0]):
    axes[i].imshow(image[i]/255.0)  # normalize RGB
    axes[i].set_title(f'Layer {i}')
    axes[i].axis('off')
fig.suptitle('3D Tensor (Image)')
plt.show()
```

```python
# 4D Tensor (Batch of Images)
batch = tf.constant([
  [[[255,0,0],[0,255,0]], [[0,0,255],[255,255,0]]],
  [[[0,255,255],[255,0,255]], [[128,128,128],[0,0,0]]]
], dtype=tf.float32)
print('4D Tensor (batch):\n', batch.numpy(), '\nShape:', batch.shape)
fig, axes = plt.subplots(1, batch.shape[0])
for i in range(batch.shape[0]):
    axes[i].imshow(batch[i]/255.0)  # normalize first image
    axes[i].set_title(f'Sample {i}')
    axes[i].axis('off')
fig.suptitle('4D Tensor (Batch of Images)')
plt.show()
```

```python
# Deep Neural Network Example (3 hidden layers)
X = np.array([[1],[2],[3],[4],[5],[6]], dtype=float)
y = np.array([[2],[4],[6],[8],[10],[12]], dtype=float)

model = tf.keras.Sequential([
    tf.keras.layers.Dense(8, activation='relu', input_shape=[1]),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(4, activation='relu'),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X, y, epochs=200, verbose=0)

preds = model.predict(X)
print('Predictions:\n', preds)

plt.plot(X, y, 'b-o', label='Actual')
plt.plot(X, preds, 'r--x', label='Predicted')
plt.legend()
plt.title('Deep Neural Network Predictions')
plt.show()
```

## TensorFlow Playground
Tweak the code below to experiment with tensors and neural networks. Change values, shapes, layers, or activations and see what happens!

```python
# Playground Example 1: Bigger 2D matrix
matrix2 = tf.constant(np.arange(16).reshape(4,4))
print('4x4 Matrix:\n', matrix2.numpy(), '\nShape:', matrix2.shape)
plt.imshow(matrix2.numpy(), cmap='plasma')
for i in range(matrix2.shape[0]):
    for j in range(matrix2.shape[1]):
        plt.text(j, i, str(matrix2.numpy()[i,j]), ha='center', va='center', color='white')
plt.show()
```

```python
# Playground Example 2: Random 3D tensor
rand_image = tf.random.uniform((3,3,3), minval=0, maxval=255, dtype=tf.float32)
plt.imshow(rand_image.numpy()/255.0)
plt.title('Random 3D Tensor')
plt.show()
```

```python
# Playground Example 3: Modify Deep NN
X_play = np.linspace(0,10,20).reshape(-1,1)
y_play = 3*X_play + np.sin(X_play)  # slightly nonlinear target

model_play = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=[1]),
    tf.keras.layers.Dense(8, activation='tanh'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1)
])
model_play.compile(optimizer='adam', loss='mean_squared_error')
model_play.fit(X_play, y_play, epochs=300, verbose=0)

preds_play = model_play.predict(X_play)
plt.plot(X_play, y_play, 'b-o', label='Actual')
plt.plot(X_play, preds_play, 'r--x', label='Predicted')
plt.legend()
plt.title('Playground NN Predictions')
plt.show()
```
